```python
fig, ax = plt.subplots(figsize = (12, 8))

valores1 = [10, 20, 30, 40]
valores2 = [20, 30, 40, 50]

etiquetas = ["a", "b", "c", "d"]

x = np.arange(len(etiquetas))

ancho = 0.5

plt.bar(x - ancho/2, valores1, width = ancho, color = "cyan")
plt.bar(x + ancho/2, valores2, width = ancho , color = "magenta")

plt.show()
```
