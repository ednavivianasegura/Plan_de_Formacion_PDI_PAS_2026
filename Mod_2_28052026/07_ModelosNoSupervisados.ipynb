```python
# =============================================
# 1. GENERACIÓN Y PREPARACIÓN DE DATOS
# =============================================

print("\n---------------------------------------------")
print('GENERACIÓN Y PREPARACIÓN DE DATOS')
print("---------------------------------------------")

# Semilla para reproducibilidad
np.random.seed(42)

# Grupo 1: personas más bajas y livianas
pesos_g1 = np.random.normal(loc=55, scale=6, size=80)     # Media 55 kg y desviación 6
estaturas_g1 = np.random.normal(loc=155, scale=6, size=80)  # Media 155 cm y desviación 6
grupo_g1 = np.full(80, 0)

# Grupo 2: personas de talla media
pesos_g2 = np.random.normal(loc=70, scale=5, size=70)     # Media 70 kg y desviación 5
estaturas_g2 = np.random.normal(loc=165, scale=5, size=70)  # Media 165 cm y desviación 5
grupo_g2 = np.full(70, 1)

# Grupo 3: personas más altas y pesadas
pesos_g3 = np.random.normal(loc=85, scale=3, size=50)     # Media 85 kg y desviación 3
estaturas_g3 = np.random.normal(loc=180, scale=3, size=50)  # Media 180 cm y desviación 3
grupo_g3 = np.full(50, 2)

# Concatenar todos los datos
pesos = np.concatenate([pesos_g1, pesos_g2, pesos_g3])
estaturas = np.concatenate([estaturas_g1, estaturas_g2, estaturas_g3])
grupos = np.concatenate([grupo_g1, grupo_g2, grupo_g3])

# Crear DataFrame Inicial
df_inicial = pd.DataFrame({
    'Peso': pesos,
    'Estatura': estaturas,
    'Grupo': grupos
})

# Combinar las dos variables en una matriz de 200x2
X = np.column_stack((pesos, estaturas))  # Cada fila es un paciente, columnas: [peso, estatura]
# np.column_stack (NumPy)tiene como propósito combinar arrays 1D o 2D de NumPy en una sola matriz 2D,
# apilándolos por columnas. Útil para unir variables numéricas antes de aplicar algoritmos de ML.
# Útil para crear una matriz numérica para algoritmos usando SCikit-learn

# ----------------------------------------------------------------------------------------------------------

# Escalado de datos (normalización)
scaler = StandardScaler()  # Crea un objeto para estandarizar (media=0, desviación=1)
X_scaled = scaler.fit_transform(X)  # Aplica la transformación a los datos


print(df_inicial.groupby('Grupo').agg({
    'Peso': ['mean', 'std'],      # Media y desviación estándar del peso
    'Estatura': ['mean', 'std', 'count']   # Media, desviación estándar de la estatura y conteo
}))
```

```python
# ==================================================
# 3. CLUSTERING JERÁRQUICO (AGGLOMERATIVE) INICIAL
# ==================================================

print("\n---------------------------------------------")
print('CLUSTERING JERÁRQUICO (AGGLOMERATIVE) INICIAL')
print("---------------------------------------------")


# Configuración del modelo
cluster_jerarquico = AgglomerativeClustering(
    #n_clusters= ,       # Por defecto 2 clusters
    metric='euclidean',  # Métrica de distancia entre puntos (distancia euclidiana)
    linkage='ward'       # Criterio para fusionar clusters:
    # linkage='ward': Minimiza la varianza de los clusters que se fusionan, produciendo clusters más compactos.
    # Alternativas comunes:
    #'complete': Usa la distancia máxima entre puntos.
    #'average': Usa la distancia promedio.
)

# Aplicar el clustering a los datos escalados
etiquetas_jer = cluster_jerarquico.fit_predict(X_scaled)  # Retorna un array con las etiquetas según el número
# de clusters creados (0 o 1)
print(etiquetas_jer)


```

```python
# --------------------------------------------
# 2. Aplicación de K-means
# --------------------------------------------

kmeans = KMeans(
    n_clusters=3,       # Número de clusters (k=3 como en el ejemplo jerárquico)
    init='k-means++',   # Método inteligente para inicializar centroides
    n_init=10,          # Número de inicializaciones diferentes (evita resultados subóptimos)
    max_iter=300,       # Máximo iteraciones por ejecución
    random_state=42     # Semilla para reproducibilidad
)
# n_clusters=3	Número de clusters (equivalente al usado en el jerárquico).
# init='k-means++'	Inicializa centroides lejanos entre sí para evitar malas convergencias.
# n_init=10	Ejecuta el algoritmo 10 veces con distintas inicializaciones y elige el mejor resultado.
# max_iter=300	Número máximo de iteraciones por ejecución.

# Entrenamiento y predicción
etiquetas_kmeans = kmeans.fit_predict(X_scaled)
centroides_kmeans = scaler.inverse_transform(kmeans.cluster_centers_)  # Centroides en escala original

print(centroides_kmeans)
print("--"*80)
print(etiquetas_kmeans)
```

```python
# @title
%%html

<link rel="stylesheet" href="https://cdn.jsdelivr.net/npm/@tabler/icons-webfont@latest/tabler-icons.min.css">
<style>
  @import url('https://fonts.googleapis.com/css2?family=Source+Code+Pro:wght@400;500&family=DM+Sans:wght@300;400;500;600&display=swap');
  .nb-cell{font-family:'DM Sans',sans-serif;padding:1.2rem 0;max-width:860px;}
  .nb-title{font-size:20px;font-weight:600;color:#1E3A8A;margin:0 0 .9rem 0;letter-spacing:-.3px;}
  .nb-divider{height:2px;background:linear-gradient(90deg,#1E3A8A 0%,#475569 60%,transparent 100%);margin-bottom:1.25rem;border:none;}
  .nb-divider-sm{height:1px;background:linear-gradient(90deg,#475569 0%,transparent 80%);margin-bottom:1rem;border:none;}
  .nb-body-box{font-size:13.5px;color:#1e293b;line-height:1.7;margin-bottom:1rem;padding:10px 14px;background:#F8FAFF;border-left:3px solid #1E3A8A;border-radius:0 6px 6px 0;}
  code.ic{font-family:'Source Code Pro',monospace;font-size:12.5px;background:#EFF6FF;color:#1E3A8A;padding:2px 7px;border-radius:4px;border:1px solid #BFDBFE;white-space:nowrap;}
  code.ic-g{font-family:'Source Code Pro',monospace;font-size:12.5px;background:#F0FDF4;color:#0F766E;padding:2px 7px;border-radius:4px;border:1px solid #99F6E4;white-space:nowrap;}
  code.ic-a{font-family:'Source Code Pro',monospace;font-size:12.5px;background:#FFF7ED;color:#92400E;padding:2px 7px;border-radius:4px;border:1px solid #FED7AA;white-space:nowrap;}
  .nb-note{display:flex;gap:10px;align-items:flex-start;background:#F0FDF4;border-left:3px solid #0F766E;border-radius:0 6px 6px 0;padding:10px 14px;margin-top:.8rem;}
  .nb-note-text{font-size:13px;color:#0f766e;line-height:1.6;}
  .warn-box{display:flex;align-items:flex-start;gap:10px;background:#FFF7ED;border-left:3px solid #F59E0B;border-radius:0 6px 6px 0;padding:10px 14px;margin-top:.8rem;}
  .warn-text{font-size:13px;color:#92400E;line-height:1.6;}
  .table-wrap{border-radius:8px;overflow:visible;border:1px solid rgba(0,0,0,.08);box-shadow:0 2px 8px rgba(0,0,0,.05);}
  .ref-table{width:100%;border-collapse:collapse;font-size:13px;}
  .ref-table thead tr{background:#1E3A8A;color:#fff;}
  .ref-table thead th{padding:10px 16px;text-align:left;font-weight:500;font-size:11px;letter-spacing:.8px;text-transform:uppercase;}
  .ref-table thead th:first-child{border-radius:8px 0 0 0;}
  .ref-table thead th:last-child{border-radius:0 8px 0 0;}
  .ref-table tbody tr:nth-child(odd){background:#F8FAFF;}
  .ref-table tbody tr:nth-child(even){background:#fff;}
  .ref-table tbody tr:hover{background:#EFF6FF;}
  .ref-table tbody td{padding:9px 16px;border-bottom:.5px solid rgba(0,0,0,.06);color:#1e293b;vertical-align:middle;}
  .ref-table tbody tr:last-child td{border-bottom:none;}
  .ref-table tbody tr:last-child td:first-child{border-radius:0 0 0 8px;}
  .ref-table tbody tr:last-child td:last-child{border-radius:0 0 8px 0;}
  .desc{font-size:13px;color:#475569;line-height:1.5;}
  .desc strong{color:#1E3A8A;font-weight:600;}
  .grid2{display:grid;grid-template-columns:1fr 1fr;gap:12px;margin-bottom:.8rem;}
  .card{background:#fff;border:1px solid rgba(0,0,0,.08);border-radius:10px;padding:1rem;}
  .card-title{font-size:11px;font-weight:600;letter-spacing:.8px;text-transform:uppercase;color:#475569;margin-bottom:.6rem;padding-bottom:.5rem;border-bottom:.5px solid rgba(0,0,0,.07);}
  .card-body{font-size:13px;color:#1e293b;line-height:1.6;}
  .hint{font-size:11px;color:#94A3B8;margin-top:.6rem;display:flex;align-items:center;gap:5px;}
  .rb{display:inline-block;font-size:10px;font-weight:600;padding:2px 7px;border-radius:20px;vertical-align:middle;white-space:nowrap;margin-left:5px;}
  .rb-blue{background:#EFF6FF;color:#1E3A8A;border:1px solid #BFDBFE;}
  .rb-teal{background:#F0FDF4;color:#0F766E;border:1px solid #99F6E4;}
  .rb-amber{background:#FFF7ED;color:#92400E;border:1px solid #FED7AA;}
</style>

<div class="nb-cell">
  <h3 class="nb-title">Selección del número óptimo de clusters</h3>
  <hr class="nb-divider">
  <div class="nb-body-box">
    Elegir incorrectamente k puede llevar a sub-segmentación (grupos heterogéneos) o sobre-segmentación
    (grupos artificiales). Dos métodos cuantitativos ayudan a tomar esta decisión.
  </div>
  <div class="grid2">
    <div class="card">
      <div class="card-title"><i class="ti ti-chart-line" style="font-size:13px;color:#1E3A8A;vertical-align:-1px;"></i> &nbsp;Método del Codo</div>
      <div class="card-body">Representa la <strong>inercia</strong> (WCSS = suma de distancias al cuadrado intra-cluster) frente a k. El punto óptimo está donde la reducción de inercia se vuelve marginal — forma de "codo" en la curva.</div>
    </div>
    <div class="card">
      <div class="card-title"><i class="ti ti-ruler" style="font-size:13px;color:#1E3A8A;vertical-align:-1px;"></i> &nbsp;Silhouette Score</div>
      <div class="card-body">Para cada punto calcula <strong>a</strong> (distancia intra-cluster) y <strong>b</strong> (distancia al cluster más cercano). El score s ∈ [−1, 1]: cercano a 1 = bien asignado · cercano a 0 = en la frontera · cercano a −1 = mal asignado.</div>
    </div>
  </div>
</div>
```

```python
# =============================================
# Configuración de rango de k's a evaluar
# =============================================

k_range = range(2, 11)  # Creamos un rango de valores para k (número de clusters) desde 2 hasta 10


# =============================================
# Cálculo de métricas
# =============================================

inercias = []          # Lista para almacenar la inercia (WCSS) para cada k
silhouette_scores = []  # Lista para almacenar los puntajes de silueta para cada k

for k in k_range:
    # Inicializamos KMeans con:
    # - n_clusters=k: número de clusters a formar
    # - random_state=42: semilla para reproducibilidad
    kmeans = KMeans(n_clusters=k, random_state=42)

    # Ajustamos el modelo y predecimos las etiquetas de cluster
    etiquetas = kmeans.fit_predict(X_scaled)

    # Almacenamos la inercia (suma de distancias al cuadrado intra-cluster)
    inercias.append(kmeans.inertia_)  # WCSS

    # Calculamos y almacenamos el silhouette score para este k
    # - X_scaled: datos estandarizados
    # - etiquetas: asignaciones de cluster predichas
    silhouette_scores.append(silhouette_score(X_scaled, etiquetas))

# Creamos una figura de 15x5 pulgadas para los subplots
plt.figure(figsize=(15, 5))

# =============================================
# Gráfico del Método del Codo (subplot izquierdo)
# =============================================
plt.subplot(1, 2, 1)  # 1 fila, 2 columnas, posición 1

# Graficamos la curva del codo:
# - k_range en eje X, inercias en eje Y
# - 'bo-': puntos azules con línea continua
# - markersize=8: tamaño de los puntos
plt.plot(k_range, inercias, 'bo-', markersize=8)

# Línea vertical roja punteada para marcar el codo sugerido
plt.axvline(x=3, color='r', linestyle='--', alpha=0.7, label='Codo sugerido (k=3)')

# Etiquetas y título
plt.xlabel('Número de clusters (k)', fontsize=12)
plt.ylabel('Inercia (WCSS)', fontsize=12)
plt.title('Método del Codo', fontsize=14)

# Configuramos ticks en el eje X para que coincidan con los valores de k probados
plt.xticks(k_range)

# Añadimos una cuadrícula semitransparente
plt.grid(alpha=0.3)

# Mostramos la leyenda
plt.legend()

# =============================================
# Gráfico del Silhouette Score (subplot derecho)
# =============================================
plt.subplot(1, 2, 2)  # 1 fila, 2 columnas, posición 2

# Graficamos los silhouette scores:
# - k_range en eje X, silhouette_scores en eje Y
# - 'go-': puntos verdes con línea continua
plt.plot(k_range, silhouette_scores, 'go-', markersize=8)

# Línea vertical roja punteada para marcar el mejor score
plt.axvline(x=3, color='r', linestyle='--', alpha=0.7, label='Mejor score (k=3)')

# Etiquetas y título
plt.xlabel('Número de clusters (k)', fontsize=12)
plt.ylabel('Silhouette Score', fontsize=12)
plt.title('Silhouette Score', fontsize=14)

# Configuramos ticks en el eje X
plt.xticks(k_range)

# Añadimos cuadrícula
plt.grid(alpha=0.3)

# Mostramos la leyenda
plt.legend()

# Ajustamos el layout para que no se solapen los elementos
plt.tight_layout()

# Mostramos la figura con ambos gráficos
plt.show()
```

```python
# =============================================
# 3. Cálculo de eigenvalores y eigenvectores
# =============================================

valores, vectores = eig(datos_scaled.cov())

vector_azul = vectores[:, 0]
vector_rojo = vectores[:, 1]

print("Eigenvector Rojo:", vector_rojo, "Eigenvalor:", valores[1])
print("Eigenvector Azul:", vector_azul, "Eigenvalor:", valores[0])

# Código de visualización

plt.figure(figsize=(7, 7))
plt.axes().set_aspect("equal")

# Graficando los datos
plt.scatter(datos_scaled["peso"], datos_scaled["estatura"],
            marker="8", s=400, color="purple", alpha=0.5)

# Graficando los eigenvectores
plt.quiver(0, 0,
           vector_azul[0]/abs(vector_azul[0])*valores[0],
           vector_azul[1]/abs(vector_azul[1])*valores[0],
           color="blue", angles="xy", scale_units="xy", scale=1, width=0.02)

plt.quiver(0, 0,
           vector_rojo[0]/abs(vector_rojo[0])*valores[1],
           vector_rojo[1]/abs(vector_rojo[1])*valores[1],
           color="red", angles="xy", scale_units="xy", scale=1, width=0.02)

plt.xlabel("peso", fontsize=15)
plt.ylabel("estatura", fontsize=15)
plt.show()
```

```python
# =============================================
# 4. Datos proyectados (nuevos ejes)
# =============================================

# ── Proyección de los datos sobre los eigenvectores ───────────────
# Multiplica los datos escalados por la transpuesta de la matriz de eigenvectores
# Esto es el núcleo matemático de PCA:
#
# datos_scaled → matriz (n_puntos × 2) con peso y estatura estandarizados
# vectores     → matriz (2 × 2) con los dos eigenvectores como filas
# vectores.T   → transpuesta (2 × 2) con eigenvectores como columnas
#
# La multiplicación matricial @ hace:
# para cada punto → calcula su coordenada en el nuevo sistema de ejes (PC1, PC2)
#
# Matemáticamente:
# X_proyectado = X_scaled · V^T
# donde V son los eigenvectores → rota el sistema de coordenadas
# alineando los ejes con las direcciones de máxima varianza
proyectados = pd.DataFrame(
    datos_scaled.values @ vectores.T,
    columns=["peso", "estatura"]   # ahora "peso" = PC1, "estatura" = PC2
)


# ── Figura ────────────────────────────────────────────────────────
plt.figure(figsize=(8, 8))

# set_aspect("equal") → misma escala en ambos ejes
# imprescindible para que los eigenvectores (perpendiculares)
# se vean realmente perpendiculares en el gráfico
plt.axes().set_aspect("equal")


# ── Datos en el nuevo sistema de coordenadas (PC1 vs PC2) ─────────
# Muestra cada punto en el espacio rotado por PCA
# PC1 (eje X) → dirección de máxima varianza (91%)
# PC2 (eje Y) → dirección de varianza residual (9%)
# Si PCA funciona bien → los puntos se dispersan más en X que en Y
plt.scatter(proyectados["peso"],     # coordenada en PC1
            proyectados["estatura"], # coordenada en PC2
            marker="8", s=400, color="purple", alpha=0.5)


# ── Proyección sobre PC1 (reducción a 1 dimensión) ───────────────
# Muestra cómo quedarían los puntos si eliminamos PC2 completamente
# → todos los puntos caen sobre el eje X (PC1) en y=-3
# Esto es literalmente lo que hace PCA al reducir de 2D a 1D:
# cada punto pierde su coordenada en PC2 y solo conserva la de PC1
#
# y=-3 → posición fija bajo el gráfico para visualizarlo claramente
# La dispersión horizontal muestra cuánta varianza conserva PC1
plt.scatter(proyectados["peso"],          # coordenada en PC1 (varía)
            [-3]*len(proyectados["estatura"]),  # todos en y=-3 (fijo)
            s=250, color="red", alpha=0.5)


# ── Proyección sobre PC2 (la dimensión que descartaríamos) ────────
# Muestra cómo quedarían los puntos si solo conservamos PC2
# → todos los puntos caen sobre el eje Y (PC2) en x=-2
# La poca dispersión vertical confirma que PC2 tiene poca varianza
# → es la dimensión que PCA descartaría al reducir a 1D
#
# x=-2 → posición fija a la izquierda del gráfico
plt.scatter([-2]*len(proyectados["estatura"]),  # todos en x=-2 (fijo)
            proyectados["estatura"],             # coordenada en PC2 (varía)
            s=250, color="blue", alpha=0.5)

plt.show()
```

```python
# =============================================
# 4. Obtención de componentes principales
# =============================================

# Crea el objeto PCA sin especificar n_components
# → calcula TODOS los componentes posibles (tantos como variables)
# En este caso 2 variables → 2 componentes principales
pca = PCA()

# fit_transform hace dos cosas en un paso:
#   fit()      → calcula los eigenvectores y eigenvalores
#                (aprende las direcciones de máxima varianza)
#   transform() → proyecta los datos sobre los nuevos ejes
#
# datos_scaled2 → matriz (n_puntos × 2)
#   columna 0 → coordenadas en PC1 (máxima varianza)
#   columna 1 → coordenadas en PC2 (varianza residual)
datos_scaled2 = pca.fit_transform(datos_scaled)


# ── Figura ────────────────────────────────────────────────────────
plt.figure(figsize=(7, 7))

# Misma escala en ambos ejes para no distorsionar las distancias
plt.axes().set_aspect("equal")

# ── Proyección sobre PC1 (puntos rojos) ───────────────────────────
# datos_scaled2[:, 0] → coordenadas de todos los puntos en PC1
# [2]*size            → todos en y=2 (línea horizontal fija)
# Muestra la dispersión de los datos proyectados sobre PC1
# Mucha dispersión → PC1 captura mucha varianza (91%)
plt.scatter(datos_scaled2[:, 0],              # coordenada en PC1 (varía)
            [2]*datos_scaled2[:, 0].size,     # todos en y=2 (fijo)
            s=250, color="red", alpha=0.5)

# ── Proyección sobre PC2 (puntos azules) ──────────────────────────
# datos_scaled2[:, 1] → coordenadas de todos los puntos en PC2
# [1]*size            → todos en y=1 (línea horizontal fija)
# Muestra la dispersión de los datos proyectados sobre PC2
# Poca dispersión → PC2 captura poca varianza (9%)
plt.scatter(datos_scaled2[:, 1],              # coordenada en PC2 (varía)
            [1]*datos_scaled2[:, 1].size,     # todos en y=1 (fijo)
            s=250, color="blue", alpha=0.5)

# Limita el eje Y para que solo se vean las dos líneas de puntos
# sin espacio vacío innecesario
plt.ylim((0.8, 2.2))
plt.show()

# ── Métricas de PCA ───────────────────────────────────────────────

# Eigenvalores → varianza absoluta capturada por cada componente
# PC1: 1.829 → captura mucha varianza
# PC2: 0.181 → captura poca varianza
print("Varianza explicada por cada componente:", pca.explained_variance_)

# Proporción de varianza → eigenvalor / suma total de eigenvalores
# PC1: 0.91 → explica el 91% de la varianza total
# PC2: 0.09 → explica el  9% de la varianza total
# Suma siempre = 1.0 (100%)
print("Proporción de varianza explicada:", pca.explained_variance_ratio_)
```

```python
# @title {"display-mode":"form"}
%%html
<!DOCTYPE html>
<html>
<head>
<style>
  .pca-wrap { padding: 1rem 0; font-family: Arial, sans-serif; }
  .controls { display: flex; gap: 8px; flex-wrap: wrap; margin-bottom: 1rem; }
  .btn {
    padding: 6px 14px; font-size: 13px; border-radius: 8px;
    border: 1px solid #ccc; background: #fff; color: #333;
    cursor: pointer; transition: background 0.15s;
  }
  .btn:hover { background: #f0f0f0; }
  .btn.active { background: #CECBF6; border-color: #7F77DD; color: #3C3489; }
  .info-row { display: grid; grid-template-columns: 1fr 1fr; gap: 12px; margin-top: 1rem; }
  .metric { background: #f5f5f5; border-radius: 8px; padding: 10px 14px; }
  .metric-label { font-size: 12px; color: #666; margin-bottom: 4px; }
  .metric-val { font-size: 18px; font-weight: 500; color: #222; }
  canvas { border-radius: 8px; border: 1px solid #ddd; display: block; width: 100%; max-width: 640px; }
  .step-label { font-size: 13px; color: #666; margin-bottom: 6px; }
  .legend { display: flex; gap: 16px; margin-top: 8px; flex-wrap: wrap; }
  .leg-item { display: flex; align-items: center; gap: 6px; font-size: 12px; color: #666; }
  .leg-dot { width: 10px; height: 10px; border-radius: 50%; }
</style>
</head>
<body>
<div class="pca-wrap">
  <div class="controls">
    <button class="btn active" onclick="setStep(0)">1. Datos originales</button>
    <button class="btn" onclick="setStep(1)">2. Ejes PCA</button>
    <button class="btn" onclick="setStep(2)">3. Solo PC1 (1D)</button>
    <button class="btn" onclick="setStep(3)">4. PC1 + PC2 (2D)</button>
  </div>
  <div class="step-label" id="step-desc">Datos originales: peso y estatura estandarizados</div>
  <canvas id="cv" height="420"></canvas>
  <div class="legend">
    <div class="leg-item"><div class="leg-dot" style="background:#7F77DD"></div>Grupo 1 (bajos/ligeros)</div>
    <div class="leg-item"><div class="leg-dot" style="background:#1D9E75"></div>Grupo 2 (talla media)</div>
    <div class="leg-item"><div class="leg-dot" style="background:#D85A30"></div>Grupo 3 (altos/pesados)</div>
  </div>
  <div class="info-row">
    <div class="metric"><div class="metric-label">Varianza PC1</div><div class="metric-val" id="var1">—</div></div>
    <div class="metric"><div class="metric-label">Varianza PC2</div><div class="metric-val" id="var2">—</div></div>
  </div>
</div>

<script>
const W = 640, H = 420, CX = W/2, CY = H/2, SC = 90;

const rng = (function() {
  let s = 42;
  return function() {
    s = (s * 1664525 + 1013904223) & 0xffffffff;
    return (s >>> 0) / 0xffffffff;
  };
})();

function randn() {
  let u = 0, v = 0;
  while (!u) u = rng();
  while (!v) v = rng();
  return Math.sqrt(-2*Math.log(u)) * Math.cos(2*Math.PI*v);
}

const groups = [
  { n: 80, mp: -0.9, me: -0.9, c: '#7F77DD' },
  { n: 70, mp:  0.0, me:  0.0, c: '#1D9E75' },
  { n: 50, mp:  0.9, me:  0.9, c: '#D85A30' }
];

let pts = [];
groups.forEach((g, gi) => {
  for (let i = 0; i < g.n; i++) {
    const p = g.mp + randn() * 0.35;
    const e = g.me + randn() * 0.35;
    pts.push({ p, e, g: gi });
  }
});

function mean(arr) { return arr.reduce((a,b)=>a+b,0)/arr.length; }
const ps = pts.map(d=>d.p), es = pts.map(d=>d.e);
const mp = mean(ps), me = mean(es);

function cov(a, b, ma, mb) {
  return a.reduce((s,v,i)=>s+(v-ma)*(b[i]-mb),0)/(a.length-1);
}
const cpp = cov(ps,ps,mp,mp), cpe = cov(ps,es,mp,me), cee = cov(es,es,me,me);

const trace = cpp+cee, det = cpp*cee-cpe*cpe;
const disc = Math.sqrt(Math.max(0,(trace/2)**2-det));
const ev1 = trace/2+disc, ev2 = trace/2-disc;
const totalVar = ev1+ev2;

let v1, v2;
if (Math.abs(cpe) > 1e-10) {
  const d1 = Math.sqrt((ev1-cee)**2+cpe**2);
  v1 = [(ev1-cee)/d1, cpe/d1];
  const d2 = Math.sqrt((ev2-cee)**2+cpe**2);
  v2 = [(ev2-cee)/d2, cpe/d2];
} else {
  v1 = [1,0]; v2 = [0,1];
}

pts.forEach(d => {
  d.pc1 = (d.p-mp)*v1[0] + (d.e-me)*v1[1];
  d.pc2 = (d.p-mp)*v2[0] + (d.e-me)*v2[1];
});

const steps = [
  { mode: 'orig' },
  { mode: 'axes' },
  { mode: 'pc1'  },
  { mode: 'both' }
];

let currentStep = 0;
let animT = 0, animId = null;

const canvas = document.getElementById('cv');
const ctx = canvas.getContext('2d');
canvas.width = W; canvas.height = H;

function toScreen(x, y) { return [CX + x*SC, CY - y*SC]; }

function drawAxes() {
  ctx.strokeStyle = 'rgba(128,128,128,0.25)';
  ctx.lineWidth = 0.5;
  ctx.beginPath(); ctx.moveTo(0,CY); ctx.lineTo(W,CY); ctx.stroke();
  ctx.beginPath(); ctx.moveTo(CX,0); ctx.lineTo(CX,H); ctx.stroke();
}

function drawEigenVectors(alpha) {
  // Longitud de cada flecha proporcional a su eigenvalor
  // PC1 tiene eigenvalor ~1.83 → flecha larga
  // PC2 tiene eigenvalor ~0.18 → flecha ~10x más corta
  const baseScale = 2.2;
  const maxEv = Math.max(ev1, ev2);

  ctx.globalAlpha = alpha;
  [[v1, ev1, '#378ADD', 'PC1 (' + (ev1/totalVar*100).toFixed(0) + '%)'],
   [v2, ev2, '#D85A30', 'PC2 (' + (ev2/totalVar*100).toFixed(0) + '%)']
  ].forEach(([v, ev, col, lbl]) => {
    // Escala proporcional al eigenvalor
    const scale = baseScale * (ev / maxEv);

    const [x1, y1] = toScreen(0, 0);
    const [x2, y2] = toScreen(v[0]*scale, v[1]*scale);

    // Línea de la flecha
    ctx.strokeStyle = col; ctx.lineWidth = 2.5;
    ctx.beginPath(); ctx.moveTo(x1,y1); ctx.lineTo(x2,y2); ctx.stroke();

    // Cabeza de flecha
    const dx = x2-x1, dy = y2-y1, len = Math.sqrt(dx*dx+dy*dy);
    const ux = dx/len, uy = dy/len;
    ctx.fillStyle = col;
    ctx.beginPath();
    ctx.moveTo(x2, y2);
    ctx.lineTo(x2 - ux*10 - uy*5, y2 - uy*10 + ux*5);
    ctx.lineTo(x2 - ux*10 + uy*5, y2 - uy*10 - ux*5);
    ctx.closePath(); ctx.fill();

    // Etiqueta junto a la punta
    ctx.fillStyle = col; ctx.font = '500 13px Arial';
    ctx.fillText(lbl, x2+8, y2+4);
  });
  ctx.globalAlpha = 1;
}

function getPtPos(d, mode) {
  if (mode === 'orig' || mode === 'axes') return [d.p, d.e];
  if (mode === 'pc1')  return [d.pc1, 0];
  if (mode === 'both') return [d.pc1, d.pc2];
}

function lerp(a, b, t) { return a + (b-a)*t; }

function draw(t) {
  ctx.clearRect(0, 0, W, H);
  drawAxes();

  const step = steps[currentStep];
  const prev = currentStep > 0 ? steps[currentStep-1] : steps[0];
  const eT = t < 0.5 ? 2*t*t : -1+(4-2*t)*t;

  if (step.mode === 'axes') drawEigenVectors(eT);
  else if (step.mode === 'pc1' || step.mode === 'both') drawEigenVectors(1 - eT*0.6);

  pts.forEach(d => {
    const [px0, py0] = getPtPos(d, prev.mode);
    const [px1, py1] = getPtPos(d, step.mode);
    const cx2 = lerp(px0, px1, eT);
    const cy2 = lerp(py0, py1, eT);
    const [sx, sy] = toScreen(cx2, cy2);
    const g = groups[d.g];
    ctx.beginPath();
    ctx.arc(sx, sy, 5, 0, Math.PI*2);
    ctx.fillStyle = g.c;
    ctx.globalAlpha = 0.65;
    ctx.fill();
    ctx.globalAlpha = 1;
    ctx.strokeStyle = g.c;
    ctx.lineWidth = 0.5;
    ctx.stroke();
  });

  if (step.mode === 'pc1') {
    ctx.strokeStyle = '#888'; ctx.lineWidth = 1.5;
    ctx.setLineDash([4,4]);
    ctx.beginPath(); ctx.moveTo(0,CY); ctx.lineTo(W,CY); ctx.stroke();
    ctx.setLineDash([]);
    ctx.fillStyle = '#444'; ctx.font = '12px Arial';
    ctx.fillText('PC1 → ' + (ev1/totalVar*100).toFixed(0) + '% varianza retenida', CX+10, CY-8);
    ctx.fillStyle = '#D85A30';
    ctx.fillText('PC2 descartada → ' + (ev2/totalVar*100).toFixed(0) + '% perdido', CX+10, CY+18);
  }

  ctx.fillStyle = '#888'; ctx.font = '11px Arial';
  ctx.fillText('peso / PC1', W-70, CY-6);
  ctx.fillText('estatura / PC2', CX+4, 16);
}

function animate() {
  animT += 0.018;
  if (animT >= 1) { animT = 1; draw(1); cancelAnimationFrame(animId); return; }
  draw(animT);
  animId = requestAnimationFrame(animate);
}

function setStep(i) {
  document.querySelectorAll('.btn').forEach((b,j) => b.classList.toggle('active', j===i));
  const descs = [
    "Datos originales: peso y estatura estandarizados",
    "Ejes PCA superpuestos: PC1 (azul) y PC2 (rojo) — longitud proporcional al eigenvalor",
    "Solo PC1 retenida: proyección sobre el primer eje (" + (ev1/totalVar*100).toFixed(0) + "% varianza)",
    "PC1 + PC2 retenidas: espacio completo rotado a los nuevos ejes"
  ];
  document.getElementById('step-desc').textContent = descs[i];
  const v1pct = (ev1/totalVar*100).toFixed(1)+'%';
  const v2pct = (ev2/totalVar*100).toFixed(1)+'%';
  if (i === 0) {
    document.getElementById('var1').textContent = '—';
    document.getElementById('var2').textContent = '—';
  } else if (i === 2) {
    document.getElementById('var1').textContent = v1pct + ' ✓ retenida';
    document.getElementById('var2').textContent = v2pct + ' ✗ descartada';
  } else {
    document.getElementById('var1').textContent = v1pct;
    document.getElementById('var2').textContent = v2pct;
  }
  currentStep = i;
  cancelAnimationFrame(animId);
  animT = 0;
  animId = requestAnimationFrame(animate);
}

setStep(0);
</script>
</body>
</html>
```

```python
# Separar características y etiqueta real
X = data.drop('Resultado', axis=1)
y = data['Resultado']  # Etiquetas reales (0: No diabetes, 1: Diabetes)

# Escalar datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Versión compatible con scikit-learn >=1.4
cluster_jerarquico = AgglomerativeClustering(
    n_clusters=2,
    metric='euclidean',  # Parámetro actualizado
    linkage='ward'
)

etiquetas_jer = cluster_jerarquico.fit_predict(X_scaled)
```

```python
# Configurar K-means
kmeans = KMeans(
    n_clusters=2,          # 2 grupos
    random_state=42        # Semilla para reproducibilidad
)

# Entrenar y predecir
etiquetas_kmeans = kmeans.fit_predict(X_scaled)
```

```python
from sklearn.metrics import adjusted_rand_score
import seaborn as sns
# Función para evaluar coincidencia con etiquetas reales
def evaluar_clusters(etiquetas_pred, y_real, metodo):
    # Matriz de confusión
    cm = confusion_matrix(y_real, etiquetas_pred)

    # Ajustar etiquetas si es necesario (K-means/jerárquico pueden invertir 0/1)
    if cm[0, 0] + cm[1, 1] < cm[0, 1] + cm[1, 0]:
        etiquetas_pred = 1 - etiquetas_pred  # Invertir etiquetas
        cm = confusion_matrix(y_real, etiquetas_pred)

    # Métricas
    ari = adjusted_rand_score(y_real, etiquetas_pred)  # Adjusted Rand Index

    # Visualización
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['No Diabetes', 'Diabetes'],
                yticklabels=['No Diabetes', 'Diabetes'])
    plt.title(f'Matriz de Confusión - {metodo}\nARI: {ari:.2f}')
    plt.show()

# Evaluar ambos métodos
evaluar_clusters(etiquetas_jer, y, 'Clustering Jerárquico')
evaluar_clusters(etiquetas_kmeans, y, 'K-means')
```

```python
# Reducir a 2D para visualización
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Gráfico comparativo
plt.figure(figsize=(15, 5))

# Clustering Jerárquico
plt.subplot(1, 2, 1)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=etiquetas_jer, cmap='viridis', alpha=0.6)
plt.title('Clustering Jerárquico (2 clusters)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')

# K-means
plt.subplot(1, 2, 2)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=etiquetas_kmeans, cmap='viridis', alpha=0.6)
plt.title('K-means (2 clusters)')
plt.xlabel('Componente Principal 1')

plt.tight_layout()
plt.show()
```
