```python
# Solución:

# ════════════════════════════════════════════════════════════════
# Paso 1 — Preparar los datos de los 4 subplots
# ════════════════════════════════════════════════════════════════

# ── Datos simulados: curvas de pérdida ──────────────────────────
np.random.seed(42)
epocas     = np.arange(1, 31)   # 30 valores: [1, 2, 3, ... 30]

# Función exponencial decreciente + ruido aleatorio pequeño
# Interpretación: conforme pasan las épocas, el error disminuye

train_loss = 1.5 * np.exp(-0.12 * epocas) + np.random.normal(0, 0.03, 30)
val_loss   = 1.5 * np.exp(-0.09 * epocas) + np.random.normal(0, 0.05, 30)

# ── Datos reales: mapeo especie → color para scatter ────────────
# Necesario porque plt.scatter(c=) requiere que cada individuo tenga asociado su color
# no una columna categórica. Se mapea cada categoría a un color.
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia los colores: 'steelblue' → 'purple', etc.        │
# └──────────────────────────────────────────────────────────────┘
color_map    = {'Iris-setosa': 'steelblue', 'Iris-versicolor': 'salmon', 'Iris-virginica': 'seagreen'}
# .map() aplica una función a cada elemento de una Serie o DataFrame. Acepta una función lambda, una función normal
# o un diccionario como parámetro, y no es in-place — devuelve un nuevo objeto.

colores_iris = iris['Clase'].map(color_map)   # Series con un color por fila

# ── Datos reales: Titanic limpio ─────────────────────────────────
titanic_clean = titanic.dropna(subset=['Age', 'Fare'])

# ════════════════════════════════════════════════════════════════
# Paso 2 — Crear la figura 2×2 y dibujar [0,0]: curvas de pérdida
# plt.plot() → línea continua, ideal para series temporales
# ════════════════════════════════════════════════════════════════

# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia figsize=(13, 9) a (10, 7) para figura más pequeña │
# │ 🟣 Cambia fontsize=15 a fontsize=10 para fuente más pequeña  │
# │ 🟣 Cambia fontsize=15 a fontsize=20 para fuente más grande  │
# │ 🟣 Escoge fontweight entre[ 'normal','bold','heavy','light', │
# │     'ultrabold','ultralight']
# └──────────────────────────────────────────────────────────────┘

fig, axes = plt.subplots(2, 2, figsize=(13, 9))
fig.suptitle('Dashboard EDA/ML — Matplotlib', fontsize=15, fontweight='bold')

# axes es un array 2D: axes[fila, columna]
# axes[0,0] → arriba izquierda
# axes[0,1] → arriba derecha
# axes[1,0] → abajo izquierda
# axes[1,1] → abajo derecha

# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia ls='--' a ':' o '-.' para la línea de val_loss   │
# │    Cambia lw=2 a 0.5 (fina) o 4 (gruesa)                   │
# └──────────────────────────────────────────────────────────────┘
axes[0,0].plot(epocas, train_loss, color='steelblue', lw=2,label='Train loss')
axes[0,0].plot(epocas, val_loss,   color='tomato', lw=2, ls='--', label='Val loss')
axes[0,0].set_title('Ej Curvas de pérdida (train vs val)')
axes[0,0].set_xlabel('Época')
axes[0,0].set_ylabel('Loss')
axes[0,0].legend()
axes[0,0].grid(True, alpha=0.1)


# ════════════════════════════════════════════════════════════════
# Paso 3 — Añadir [0,1]: scatter coloreado por clase
# Objetivo: ver si las especies son visualmente separables
# (mismo análisis que haría un modelo de clasificación)
# ════════════════════════════════════════════════════════════════

# [0,1] scatter con c= (lista de colores, uno por punto)
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia alpha=0.6 a 0.2 (muy transparente) o 1.0 (opaco) │
# │    Cambia s=30 a 80 → puntos más grandes                    │
# │    Cambia 'Largo Sepalo' a 'Largo Petalo' (mejor separación)│
# └──────────────────────────────────────────────────────────────┘
axes[0,1].scatter(iris['Largo Sepalo'], iris['Ancho Petalo'],
                  c=colores_iris,           # lista de colores por punto
                  alpha=0.6, s=30, edgecolors='white', lw=0.4)

# Leyenda manual (scatter con c= no genera leyenda automática)
leyenda = [Patch(color=c, label=e) for e, c in color_map.items()]
axes[0,1].legend(handles=leyenda, fontsize=9)
axes[0,1].set_title('Scatter: separabilidad por clase')
axes[0,1].set_xlabel('Largo Sepalo')
axes[0,1].set_ylabel('Ancho Sepalo')


# ════════════════════════════════════════════════════════════════
# Paso 4 — Añadir [1,0] hist
# ════════════════════════════════════════════════════════════════

# [1,0] Histogramas superpuestos — distribución de una caracterísitca por grupo
# Patrón: df[df['col']==valor]['feature'] filtra el DataFrame por grupo
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia 'Fare' a 'Age' para ver la distribución de edades │
# │    Cambia bins=10 a 5 (menos detalle) o 20 (más)           │
# │    Cambia alpha=0.45 a 0.8 para ver mejor cada barra       │
# └──────────────────────────────────────────────────────────────┘
for clase, color in [(1,'blue'), (2,'red'), (3,'green')]:
    axes[1,0].hist(titanic[titanic['Pclass'] == clase]['Fare'],
                   bins=10, alpha=0.45, color=color, label=f'{clase}ª clase')
axes[1,0].set_title('Distribución del precio del billete por clase')
axes[1,0].set_xlabel('Fare ($)')
axes[1,0].legend()

# ════════════════════════════════════════════════════════════════
# Paso 5 — Añadir [1,1] boxplot → figura completa
# ════════════════════════════════════════════════════════════════

# [1,1] Boxplot — mediana + IQR + outliers para cada grupo
# datos_box: lista de arrays → una caja por array
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia notch=True a False para quitar la muesca          │
# │    Cambia patch_artist=True a False → cajas sin relleno     │
# └──────────────────────────────────────────────────────────────┘
datos_box = [titanic[titanic['Pclass'] == c]['Age'].dropna().values for c in [1, 2, 3]]
bp = axes[1,1].boxplot(datos_box, tick_labels=['1ª', '2ª', '3ª'],
                        patch_artist=True, notch=True)
for patch, col in zip(bp['boxes'], ['#3B82F6', '#10B981', '#F59E0B']):
    patch.set_facecolor(col)  ;  patch.set_alpha(0.6)
axes[1,1].set_title('Distribución de edad por clase')
axes[1,1].set_ylabel('Edad')

plt.tight_layout()
plt.savefig('dashboard_mpl.png', dpi=300, bbox_inches='tight')  # ← SIEMPRE antes de show()
plt.show()

```

```python
# ════════════════════════════════════════════════════════════════
# PASO 1 — Desbalanceo de clases con countplot()
# ¿Hay el mismo número de ejemplos por clase?
# ════════════════════════════════════════════════════════════════
sns.set_theme(style='whitegrid')

# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia palette='Set2' a 'pastel', 'deep', 'muted'       │
# │    Con hue='species' cada barra tiene su propio color       │
# └──────────────────────────────────────────────────────────────┘
sns.countplot(data=iris, x='Clase', hue='Clase', palette='Set2', legend=False)
plt.title('Paso 1 — Balanceo de clases (countplot)')
plt.xlabel('Especie')
plt.ylabel('Número de muestras')
plt.show()
print('Conteo:', iris['Clase'].value_counts().to_dict())
```

```python
# ════════════════════════════════════════════════════════════════
# PASO 2 — Matriz de correlación con heatmap()
# Features muy correlacionadas → candidatas a eliminar (redundantes)
# ════════════════════════════════════════════════════════════════
corr = iris.select_dtypes('number').corr()
print("--"*80)
print('Correlación entre features:')
print(corr)
print("--"*80)


# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia cmap='coolwarm' a 'viridis', 'RdBu', 'Blues'     │
# │    Elimina vmin=-1, vmax=1 → escala libre (rango real)      │
# │    Cambia annot=True → que no se vean las correlaciones     │
# └──────────────────────────────────────────────────────────────┘
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            linewidths=0.5, vmin=-1, vmax=1)
plt.title('Paso 2 — Correlación entre features (heatmap)')
plt.show()
print('Largo Petalo y Ancho Petalo tienen correlación ≈ 0.96 → altamente correlacionadas')
```

```python
# ════════════════════════════════════════════════════════════════
# PASO 3 — Separabilidad de clases con pairplot()
# Si las clases se separan visualmente → un clasificador puede funcionar
# ════════════════════════════════════════════════════════════════

# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia diag_kind='hist' a 'kde'                         │
# │    Añade corner=True para ver solo la mitad inferior        │
# └──────────────────────────────────────────────────────────────┘
g = sns.pairplot(iris, hue='Clase', diag_kind='hist',
                 palette='Set2', corner=False)
g.figure.suptitle('Paso 3 — Separabilidad de clases (pairplot)', y=1.02) #y=1.02 controla la posición vertical del título de la figura.
plt.show()
```

```python
# ════════════════════════════════════════════════════════════════
# PASO 4 — Distribución de features por clase: boxplot + histplot
# ════════════════════════════════════════════════════════════════
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# [0] boxplot — mediana + IQR + outliers por especie
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia y='petal_length' a 'sepal_width' u otra feature  │
# │    Cambia palette='Set1' a 'deep', 'pastel', 'muted'       │
# └──────────────────────────────────────────────────────────────┘
sns.boxplot(data=iris, x='Clase', y='Largo Petalo',
            hue='Clase', palette='Set1', ax=axes[0], legend=False)

axes[0].set_title('boxplot — Largo Petalo por especie')

# [1] histplot con KDE — distribución continua por especie
# ┌──────────────────────────────────────────────────────────────┐
# │ 🟣 Cambia kde=True a False para quitar la curva de densidad │
# │    Cambia 'Largo Petalo' a otra columna numérica            │
# └──────────────────────────────────────────────────────────────┘
sns.histplot(data=iris, x='Largo Petalo', hue='Clase',
             bins=20, kde=True, palette='Set2', ax=axes[1])
axes[1].set_title('histplot(kde=True) — Largo Petalo por especie')

plt.suptitle('Paso 4 — Distribución de features por clase', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```
